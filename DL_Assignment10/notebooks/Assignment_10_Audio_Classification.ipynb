## 6. Interpretation of metrics

- **Accuracy:** overall proportion of correctly classified environmental sounds.
- **Macro precision:** class-wise precision averaged equally across the 50 classes.
- **Macro recall:** class-wise recall averaged equally across the 50 classes.
- **Macro F1-score:** class-wise F1 averaged equally across classes.
- **Confusion matrix:** highlights classes that the CNN often mixes up.
- **STT WER:** Word Error Rate can be calculated only when you have a reference transcript for each speech sample. ESC-50 is not a speech-transcription dataset, so the project does not claim a WER result by default.
